# Getting started with comtradeapicall

Simple, direct examples for learning the library. Run the setup cells first, then choose an example. Each example calls `comtradeapicall` directly and displays its result.

**Start with public preview examples.** Authenticated examples need your own subscription key. Download examples create files and async examples submit jobs: run them individually, not with Run All.

This is a learning notebook, not a regression test suite. Examples cover all 35 public functions exposed by the supplied package. No credentials or saved API outputs are included.

## 1. Import the library
Use a Python environment where your chosen version of the library is installed, or put this notebook alongside your local `comtradeapicall/` folder. The printed path shows which copy Python imported.

```python
from pathlib import Path
import os
import comtradeapicall

print("Library loaded from:", comtradeapicall.__file__)
```

## 2. Optional subscription key and download folder
Public preview examples do not require a key. For authenticated examples, set `SUBSCRIPTION_KEY` in your environment or a private `.env` file. Never share the real key or your `.env` file. A missing key is not a reason to run authenticated examples; skip those sections until configured.

```python
# Optional: load a private .env file if python-dotenv is installed.
try:
    from dotenv import load_dotenv
    load_dotenv(Path.cwd() / ".env", override=False)
except ImportError:
    pass

subscription_key = (os.getenv("SUBSCRIPTION_KEY") or "").strip() or None
directory = str(Path("./comtrade_downloads"))
Path(directory).mkdir(parents=True, exist_ok=True)
print("Subscription key configured:", subscription_key is not None)
```

## Example selections
Most examples use reporter `36` (Australia), monthly period `202205` (May 2022), and `HS` classification. `M` means imports and `X` means exports. Edit the visible arguments to change your selection.

`None` usually leaves a filter unspecified. Keep reporter and period selections bounded when learning, especially for downloads. Empty results can be legitimate. Availability examples below omit recent-publication filters so they are not tied to the day you run the notebook.

## 8. Asynchronous requests: run individually

### submitAsyncFinalDataRequest


```python
final_request = comtradeapicall.submitAsyncFinalDataRequest(
    subscription_key,
    typeCode='C',
    freqCode='M',
    clCode='HS',
    period='202205',
    reporterCode='36',
    cmdCode='91,90',
    flowCode='M',
    partnerCode=None,
    partner2Code=None,
    customsCode=None,
    motCode=None,
    aggregateBy=None,
    breakdownMode='classic',
)
print("Request ID:", final_request["requestId"])
```

### submitAsyncTarifflineDataRequest


```python
tariffline_request = comtradeapicall.submitAsyncTarifflineDataRequest(
    subscription_key,
    typeCode='C',
    freqCode='M',
    clCode='HS',
    period='202205',
    reporterCode='36',
    cmdCode='91,90',
    flowCode='M',
    partnerCode=None,
    partner2Code=None,
    customsCode=None,
    motCode=None,
)
print("Request ID:", tariffline_request["requestId"])
```

### checkAsyncDataRequest
Run the final-data submission cell first. Rerun this status cell later if the request is still processing.

```python
df = comtradeapicall.checkAsyncDataRequest(
    subscription_key,
    batchId=final_request['requestId'],
)
df.head()
```
